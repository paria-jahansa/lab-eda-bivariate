```python
import kagglehub
from pathlib import Path
path = kagglehub.dataset_download("asaniczka/uk-optimal-product-price-prediction")
files = list(Path(path).glob("*.csv"))
files
```

```python
data = pd.read_csv(files[0])
df=data.copy()
df
```

```python
#Part 1-1
tab=pd.crosstab(df.isBestSeller,df.category)
tab
```

```python
ratios = tab.loc[True] / tab.loc[False]
ratios[ratios>1]
```

Based on the results, there is no category in which best-selling products are more prevalent than non-best-selling products.


```python
#Part1-2
from scipy.stats import chi2_contingency
_, chi2_pvalue, _, _  = chi2_contingency(tab)
chi2_pvalue
```

Because the p-value is less than 0.05, we reject the null hypothesis that best-seller status and product category are independent. This suggests that there is a statistically significant association between product category and whether a product is a best seller.


```python
from scipy.stats.contingency import association
association(tab, method='cramer')
```

There is a statistically significant association between product category and best-seller status; however, Cramér’s V of 0.122 indicates that the strength of this association is weak.

```python
#Part1-3
tab.plot(
    kind="bar",
    stacked=True,
    figsize=(12, 6)
)

plt.legend(
    bbox_to_anchor=(1.05, 1),
    loc="upper left",
    ncol=6
)

plt.tight_layout()
plt.show()
```

```python
#Part2-1
def outlier_slayer(data):
    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1

    # Define bounds for the outliers
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identify the outliers
    outliers = data[(data < lower_bound) | (data > upper_bound)]

    return outliers


rows_to_remove=outlier_slayer(df.price)
rows_to_remove.
```

```python
df_filtered = df.drop(rows_to_remove.index)
df_filtered
```
